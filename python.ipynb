```python
import matplotlib.pyplot as plt
import numpy as np
import math


def serie_harmonique():
    inv = 1/np.arange(1, 100_000)
    return [np.sum(inv[:n]) for n in range(1, 100_001)]

# Plotting the logarithmic divergence and harmonic series
x = np.arange(1, 100_001)
plt.plot(x, np.log(x))
plt.plot(x, serie_harmonique())
plt.xlabel("n")
plt.ylabel("H(n)")
plt.title("Divergence logarithmique et constante d'Euler")
plt.show()
```
